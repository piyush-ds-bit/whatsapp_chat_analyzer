```python
from dateutil import parser

# 1️⃣  ── remove the trailing " - " (or " - Name:")
df['clean_ts'] = (
    df['message_date']
      .astype(str)
      .str.extract(r'^(.*?)(?:\s+-\s+|$)', expand=False)   # keep only the timestamp
      .str.strip()
)

# 2️⃣  ── helper that tries DD/MM first, then MM/DD
def parse_any(ts: str):
    for dayfirst in (True, False):        #  loop twice → fast & reliable
        try:
            return parser.parse(ts, dayfirst=dayfirst)
        except (ValueError, OverflowError):
            continue
    return pd.NaT                          # if it still fails, mark as missing

# 3️⃣  ── parse everything
df['Date'] = pd.to_datetime(df['clean_ts'].map(parse_any), errors='coerce')

# 4️⃣  ── drop the temporary column if you like
df.drop(columns=['message_date', 'clean_ts'], inplace=True)

# 5️⃣  ── the usual time-based features
df['Year']   = df['Date'].dt.year
df['Month']  = df['Date'].dt.month_name()
df['Day']    = df['Date'].dt.day
df['Hour']   = df['Date'].dt.hour
df['Minute'] = df['Date'].dt.minute
```

```python
port_stem = PorterStemmer()
```

```python
def stemming(messages):

    stemmed_messages=re.sub('[^a-zA-Z]',' ',messages)
    stemmed_messages=stemmed_messages.lower()
    stemmed_messages=stemmed_messages.split()
    stemmed_messages=[port_stem.stem(word) for word in stemmed_messages if not word in stopwords.words('english') and stop_words]
    stemmed_messages=' '.join(stemmed_messages)
    return stemmed_messages
```

```python
def personality_one_liner(selected_user,df):
    if selected_user != 'OverAll':
        df = df[df['Users'] == selected_user]

    total_msgs = df.shape[0]
    total_words = df['Messages'].astype(str).apply(lambda x: len(x.split())).sum()
    media_msgs = df['Messages'].astype(str).str.contains('<Media omitted>').sum()
    emoji_count = df['Messages'].astype(str).str.count(r'[^\w\s,]').sum()
    question_marks = df['Messages'].astype(str).str.count(r'\?').sum()
    caps_count = df['Messages'].astype(str).apply(lambda x: x.isupper()).sum()
    avg_msg_len = total_words / total_msgs if total_msgs > 0 else 0
    night_msgs = df[df['Hour'].between(0, 5)].shape[0]
    if media_msgs / total_msgs > 0.3:
        return "Media Spammer 📸"
    elif emoji_count / total_msgs > 1:
        return "The Emoji King 🤴"
    elif night_msgs / total_msgs > 0.5:
        return "Midnight Philosopher 🌙"
    elif question_marks / total_msgs > 0.2:
        return "The Question Machine ❓"
    elif caps_count / total_msgs > 0.1:
        return "Caps Lock Commander 🔊"
    elif avg_msg_len > 25:
        return "The Storyteller 📖"
    elif total_msgs < 20:
        return "Silent Observer 👀"
    else:
        return "Just Vibin’ 😎"
```
